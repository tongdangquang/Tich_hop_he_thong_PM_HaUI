```python
import requests
url = "http://localhost/api/api/th/get_sinhvien"
```

```python
try:
    response = requests.get(url)
    response.raise_for_status()

    data = response.json()  # Trả về list các dict/json

    print("DANH SACH SINH VIÊN: ")
    print(f"{'MÃ SV':<10}{'HỌ TÊN':<25}{'ĐỊA CHỈ':<25}{'ĐIỆN THOẠI':<15}{'ẢNH':<10}{'MÃ LỚP':<10}")
    for sv in data:
        print(f"{sv['masv']:<10}{sv['hoten']:<25}{sv['diachi']:<25}{sv['dienthoai']:<15}{sv['anh']:<10}{sv['malop']:<10}")

except requests.exceptions.RequestException as e:
    print("Lỗi khi gọi API:", e)
```
